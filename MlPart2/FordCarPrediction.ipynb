```python
x=df.drop('price',axis=1)
y=df['price']
```

```python
df.columns
```

```python
x_one_encoded=pd.get_dummies(x,columns=['model','transmission','fuelType'])
```

```python
x_one_encoded
```

```python
x_one_encoded=x_one_encoded.astype(int)
```

```python
x_one_encoded
```

```python
from sklearn.preprocessing import LabelEncoder
encode=LabelEncoder()
columns=['model','transmission','fuelType']
xlabel=x
for i in columns:
    xlabel[i]=encode.fit_transform(xlabel[i])
```

```python
xlabel
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numerical_cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize']

scaler = StandardScaler()

x_one_encoded[numerical_cols] = scaler.fit_transform(
    x_one_encoded[numerical_cols]
)
```

```python
x_one_encoded
```

```python
xlabel[['model', 'year', 'transmission', 'mileage', 'fuelType',
        'tax', 'mpg', 'engineSize']] = scaler.fit_transform(
            xlabel[['model', 'year', 'transmission', 'mileage',
                    'fuelType', 'tax', 'mpg', 'engineSize']]
        )
```

```python
xlabel
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x_one_encoded,
    y,
    test_size=0.33,
    random_state=42
)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
y_pred
```

```python
y_test
```

```python
#comparing ytest and y_pred
```

```python
r2=r2_score(y_test,y_pred)
```

```python
r2
```

```python
n = x_test.shape[0]
p = x_test.shape[1]

adjusted_r2 = 1 - ((1 - r2) * (n - 1)) / (n - p - 1)

print("Adjusted R² Score:", adjusted_r2)
```
